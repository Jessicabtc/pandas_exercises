```python
import pandas as pd

# 1️⃣ 创建原始长格式 DataFrame
data = {
    'City': ['NY', 'NY', 'SF', 'SF', 'LA', 'LA'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [10, 20, 30, 40, 50, 60]
}

df = pd.DataFrame(data)
print("原始长格式 DataFrame:")
print(df)
```

```python
df1 = df.groupby(['City','Product'])['Sales'].sum()
df1 #聚合结果（长格式 + MultiIndex)
```

```python
df2 = df1.unstack() #宽格式
df2
```

```python
df2['A']+df2['B']
```
